```python
filling_columns = ["Mg","K"]
```

```python
df_new = df.copy()
```

```python
for col in filling_columns:
    df_new[col] = df[col].replace(0,np.nan)
```

```python
df_new.isnull().sum()
```

```python
df_new = df_new.drop(["Fe","Ba"],axis=1)
```

```python
df_new["Mg"] = df_new["Mg"].fillna(df_new["Mg"].mean())
```

```python
df_new["K"] = df_new["K"].fillna(df_new["K"].mean())
```

```python
df_new.isnull().sum()
```

```python
df_new.describe()
```

```python
sns.heatmap(df_new.corr(),annot=True)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df_new.drop("Type",axis=1)
y = df_new["Type"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=15,stratify=y)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
y_train = le.fit_transform(y_train)
y_test = le.fit_transform(y_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix, classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier,AdaBoostClassifier,GradientBoostingClassifier
import xgboost as xgb
import lightgbm as lgb
from sklearn.naive_bayes import GaussianNB

def evulate_model(true,predicted):
    score = accuracy_score(true,predicted)
    matrix = confusion_matrix(true,predicted)
    report = classification_report(true,predicted)

    return score,matrix,report

models = {
    "Logistic Regression": LogisticRegression(),
    "K Neighbors Classifier": KNeighborsClassifier(),
    "SVC": SVC(),
    "Decision Tree Classifier": DecisionTreeClassifier(),
    "Random Forest Classifier": RandomForestClassifier(),
    "AdaBoost Classifier": AdaBoostClassifier(),
    "Gradient Boost Classifier": GradientBoostingClassifier(),
    "Gaussian NB": GaussianNB(),
    "XGBoost Classifier": xgb.XGBClassifier(),
    "LightGBM CLassifier": lgb.LGBMClassifier()
}

for i in range(len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train,y_train)

    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    model_train_score,model_train_matrix,model_train_report = evulate_model(y_train,y_train_pred)
    model_test_score,model_test_matrix,model_test_report = evulate_model(y_test,y_test_pred)

    print(list(models.keys())[i])
    print("Model performance for Traning set")
    print("Score",model_train_score)
    print("Matrix",model_train_matrix)
    print("Report",model_train_report)

    print("---------------------------------")

    print("Model performance for Test set")
    print("Score",model_test_score)
    print("Matrix",model_test_matrix)
    print("Report",model_test_report)
    print("\n")
```

```python
#HYPERPARAMETER TUNING
# AFTER HERE ALL CODES MADE BY AI
```

```python
import warnings
warnings.filterwarnings('ignore') # Suppress LightGBM/Sklearn zero-division warnings for clean output

# 1. Initialize models with regularized hyperparameters to prevent overfitting
models = {
    "Logistic Regression": LogisticRegression(class_weight='balanced', max_iter=1000),
    "K Neighbors Classifier": KNeighborsClassifier(n_neighbors=5), # slightly higher k can reduce overfitting
    "SVC": SVC(class_weight='balanced'),
    "Decision Tree Classifier": DecisionTreeClassifier(max_depth=4, min_samples_split=10, class_weight='balanced'),
    "Random Forest Classifier": RandomForestClassifier(n_estimators=100, max_depth=5, min_samples_split=10, class_weight='balanced'),
    "AdaBoost Classifier": AdaBoostClassifier(n_estimators=50, learning_rate=0.1), # lower learning rate
    "Gradient Boost Classifier": GradientBoostingClassifier(max_depth=3, min_samples_split=10, learning_rate=0.05),
    "Gaussian NB": GaussianNB(),
    "XGBoost Classifier": xgb.XGBClassifier(max_depth=3, learning_rate=0.05, subsample=0.8, colsample_bytree=0.8),
    "LightGBM Classifier": lgb.LGBMClassifier(max_depth=3, num_leaves=7, learning_rate=0.05, min_child_samples=10, verbose=-1)
}

# 2. Execute and output nicely
for name, model in models.items():
    # Fit the model
    model.fit(X_train, y_train)

    # Make predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # Evaluate
    train_score, train_matrix, train_report = evulate_model(y_train, y_train_pred)
    test_score, test_matrix, test_report = evulate_model(y_test, y_test_pred)

    # Output formatting
    print("=" * 60)
    print(f"🚀 MODEL: {name}")
    print("=" * 60)
    
    print("--- 📊 TRAINING SET PERFORMANCE ---")
    print(f"Accuracy Score: {train_score:.4f}\n")
    print("Confusion Matrix:\n", train_matrix, "\n")
    print("Classification Report:\n", train_report)
    
    print("-" * 60)
    
    print("--- 🎯 TEST SET PERFORMANCE ---")
    print(f"Accuracy Score: {test_score:.4f}\n")
    print("Confusion Matrix:\n", test_matrix, "\n")
    print("Classification Report:\n", test_report)
    print("\n\n")
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import cross_val_score, StratifiedKFold
import numpy as np

# 1. Encode the full target array
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# 2. Setup a 5-fold stratified cross-validator
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

print("=" * 60)
print("🔄 5-FOLD CROSS-VALIDATION RESULTS")
print("=" * 60)

# 3. Iterate through your existing 'models' dictionary
for name, model in models.items():
    
    # Use y_encoded instead of y
    cv_scores = cross_val_score(model, X, y_encoded, cv=cv, scoring='accuracy', n_jobs=-1)
    
    # Output the results
    print(f"🚀 {name}")
    print(f"Individual Fold Scores: {np.round(cv_scores, 3)}")
    print(f"Mean Accuracy: {cv_scores.mean():.4f} (± {cv_scores.std():.4f})")
    print("-" * 60)
```

```python
import lightgbm as lgb
import pickle
from pathlib import Path

best_model = lgb.LGBMClassifier(
    max_depth=3,
    num_leaves=7,
    learning_rate=0.05,
    min_child_samples=10,
    verbose=-1
)

best_model.fit(X, y_encoded)

models_path = Path("../datasets/models")
models_path.mkdir(parents=True, exist_ok=True)

with open(models_path / "classifier.pkl", "wb") as file:
    pickle.dump(best_model, file)

# LabelEncoder'ı kaydet
with open(models_path / "label_encoder.pkl", "wb") as file:
    pickle.dump(le, file)

# Scaler'ı kaydet
with open(models_path / "scaler.pkl", "wb") as file:
    pickle.dump(scaler, file)

print("Model, Scaler and LabelEncoder saved succesfully!")
```
